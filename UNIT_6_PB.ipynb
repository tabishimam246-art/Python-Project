Example: If value in text file is : ‘INDIA IS MY COUNTRY’

```python
%%writefile Story.txt
INDIA IS MY COUNTRY
```

```python
# Q14: write your solution here
f = open("Story.txt")
words = f.read().split()
for i in words:
    if i[0] == "I":
        print(i[::-1], end = " ")
    else:
        print(i, end = " ")
```

```python
# Q16: write your solution here
f = open("Story2.txt")
n4 = open("new4.txt", "w+")
lines = f.readlines()
for i in lines:
    if i[0] == "#":
        continue
    elif i.find("#") != -1:
        n4.write(i[:i.find("#")])
    else:
        n4.write(i)
n4.seek(0)
print(n4.read())
```

```python
# Q18: write your solution here
f1 = open("python1.txt")
f2 = open("python2.txt")
l1 = f1.readlines()
l2 = f2.readlines()
for ln, line in enumerate(l1):
    if line != l2[ln]:
        for col, char in enumerate(line):
            if char != l2[ln][col]:
                print("Line Number:", ln+1, "Col Number:", col+1)
```

# Write a program that prompts for a file name, then opens that file and reads through the file, looking for lines of the form: X-DSPAM-Confidence: 0.8475 Count these lines and extract the floating point values from each of the lines and compute the average of those values and produce an output as shown below. Do not use the sum() function or a variable named sum in your solution. You can download the sample data at http://www.py4e.com/code3/mbox-short.txt when you are testing below enter mbox-short.txt as the file name

```python
f = open("mbox-short (2).txt")
count = 0
value = 0
for i in f:
    if i.startswith("X-DSPAM-Confidence: "):
        value += float(i.split()[1])
        count += 1
print(value / count)        
```
