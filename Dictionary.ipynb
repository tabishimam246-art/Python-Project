2. datatype of value can be anything

```python
d = {
    1:10,
    2:8,
    3:10,
    4:8,
    5:10
}
d[7] = 10
print(d)
print(d.get(4))
print(d.get(2, 10))
print(d)
```

```python
s = "aadsasdbssdddhhhhhh"
d = {}
for i in s:
    d[i] = d.get(i, 0) + 1 # we have to increase value not key therefore d[i] += 1 will give error as it tries to increase the key which is immutable
print(d)    
```

```python
s = "apple orange banana ant an ball bat cow cat an dog deer don fish fall"
d = {}
l = s.split()
for i in l:
    d[i[0]] = d.get(i[0], [])+[i]
print(d)    
```

```python
# 2nd method
for i in l:
    if i[0] not in d:
        d[i[0]] = [i]
    else:
        d[i[0]].append(i)
print(d)        
```

```python
# 1 - one
# 55 - fifty five
# 785 - seven hundred eighty five
# n = str(input("Enter a number : "))
def figure_to_words(n):
    d = {
        1 : "One",
        2 : "Two",
        3 : "Three",
        4 : "Four",
        5 : "Five",
        6 : "Six",
        7 : "Seven",
        8 : "Eight",
        9 : "Nine",
        10 : "Ten",
        11 : "Eleven",
        12 : "Twelve",
        13 : "Thirteen",
        14 : "Fourteen",
        15 : "Fifteen",
        16 : "Sixteen",
        17 : "Seventeen",
        18 : "Eighteen",
        19 : "Nineteen",
        20 : "Twenty",
        30 : "Thirty",
        40 : "Forty",
        50 : "Fifty", 
        60 : "Sixty",
        70 : "Seventy",
        80 : "Eighty", 
        90 : "Ninety",
        100 : "Hundred"
    }
    if n <= 20:
        return d[n]
    elif n < 100:
        if n % 10 == 0:
            return d[n]
        else:
            return d[(n//10)*10] + " " + d[n%10]
    elif n < 1000:
        if n % 100 == 0:
            return d[n//100]+" Hundred"
        else:
            return d[(n//100)]+" Hundred"+" "+figure_to_words(n%100)
print(figure_to_words(109))          
```

### len, min, max, max, sorted, reversed [Only for keys]

```python
d = {1: 2, 3: 4, 5: 6, 7: 8, 2: 10}
print(len(d))
print(max(d))
print(min(d))
print(sorted(d))
print(list(reversed(d)))
```

```python
# to find second highest frequency
s = "aadsasdbssdddhhhhhh"
d = {}
for i in s:
    d[i] = d.get(i, 0) + 1 # we have to increase value not key therefore d[i] += 1 will give error as it tries to increase the key which is immutable
l = d.items()

sorted(l, key = lambda x:x[-1], reverse = True)[1]
```
