```python
t = (1, 2, 3)
print("*"*20)
for i in t:
    print(i)
print("$"*20)
for i in range(len(t)):
    print(i, "index", t[i], "value")
print("_"*20)
for i, j in enumerate(t):
    print(i, "index", j, "value")
print("^"*20)
print(enumerate(t))
print(tuple(enumerate(t)))
print(tuple(enumerate(t, start = 10))) # index will start from 10
```

```python
t = ((1, 10), (10, 10, 1), (7, 10), (10,))
target = 10
ans = ()
for i, j in enumerate(t):
    for k in range(len(j)):
        if j[k] == target:
            ans += ((i, k),)
print(ans)            
```

```python
# 1D
t = (12, 32, 9, 67, 1)
min = t[0]
max = t[0]
for i, j in enumerate(t):
    if j > max:
        max = j
    elif j < min:
        min = j
print(max, min)        
```

```python
# 2D
t = ((12, 32), (9, 67), (1,))
maximum = t[0][0]
minimum = t[0][0]
for i, j in enumerate(t):
    for k in range(len(j)):
        if j[k] > maximum:
            maximum = j[k]
        elif j[k] < minimum:
            minimum = j[k]
print(maximum, minimum)        
```
