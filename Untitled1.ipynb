```python
from functools import reduce 
  
records = [ 
    ("b", 4), ("a", 3), ("b", 2), ("a", 3), ("c", 5) 
] 
  
selected = list(filter(lambda row: row[1] >= 3, records) )
print(selected)
ordered = sorted(selected, key=lambda row: (row[0], -row[1])) 
print(ordered)
result = reduce(lambda acc, row: acc + [row[1]], ordered, []) 
  
print(result) 
```

```python
pairs = [("x", 2), ("y", 3), ("x", 5), ("z", 4), ("y", 7)] 
  
result = { 
    key: (value, key * value) 
    for key, value in pairs 
    if value % 2 
} 
  
print(list(result.items())) 
```

```python
from functools import reduce 
  
source = iter([2, 5, 8, 11, 14]) 
evens = list(filter(lambda n: n % 2 == 0, source)) 
total = reduce(lambda acc, n: acc + n, source, 100) 
  
print(evens, total, list(source)) 
```
