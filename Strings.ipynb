# count, find, rfind, index, rindex

```python
s = "The rain in spain"
print(s.count("a"))
print(s.count("in"))
print(s.count("z"))
print(s.count("in", 7, 16))
print(s.count("", 7, 16))
```

```python
s = "The rain in spain"
print(s.find("a")) # index of first occurence of substring
print(s.rfind("a")) # index of last occurence of substring
print(s.find("a", 6, 18))
print(s.rfind("in", 1, 8))
print(s.find("kj"))
print(s.rfind("The", 2, 18))
```

```python
s = "The rain in spain"
print(s.index("a")) # index of first occurence of substring
print(s.rindex("a")) # index of last occurence of substring
print(s.index("a", 6, 18))
print(s.rindex("in", 1, 8))
print(s.index("kj")) # it wont give -1, it will give value error
print(s.rindex("The", 2, 18))
```

```python
# Encryption Logic
s1 = str(input("Enter String : "))
key = int(input("Enter key : "))
enc = ""
cap = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
number = "0123456789"
low = cap.lower()
for i in s1:
    if i in cap:
        x = cap.find(i)
        enc += cap[(x+key)%26]
    elif i in low:
        x = low.find(i)
        enc += low[(x+key)%26]
    elif i in number:
        x = number.find(i)
        enc += number[(x+key)%10]
    else:
        enc += i
print(enc)    
```

```python
# Decryption Logic
s1 = str(input("Enter String : "))
key = int(input("Enter key : "))
dec = ""
cap = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
number = "0123456789"
low = cap.lower()
for i in s1:
    if i in cap:
        x = cap.find(i)
        dec += cap[x-key]
    elif i in low:
        x = low.find(i)
        dec += low[x-key]
    elif i in number:
        x = number.find(i)
        dec += number[x-key]
    else:
        dec += i
print(dec)
```

### Password validation
1. Length greater than 7
2. Atleast 1 upper character
3. Atleast 1 lower character
4. Atleast 1 special character
5. Atleast 1 digit
6. No space allowed
   
Give the message if any condition not satisfy otherwise display "valid password" 

```python
valid = False
while not valid:
    password = str(input("Enter password\n1. Length greater than 7 \n2. Atleast 1 upper character \n3. Atleast 1 lower character \n4. Atleast 1 special character \n5. Atleast 1 digit \n6. No space allowed \n "))
    valid_len = upper = lower = special = digit = space = False
    if len(password) > 7:
        valid_len = True
    elif password.isupper():
        upper = True
    elif password.islower():
        lower = True
    elif password.isdigit():
        digit = True
    elif password.isspace():
        space = True
    else:
        special = True

    if not valid_len:
        print("Length of password must be greater than 7")
        continue
    if not upper:
        print("One upper character is required")
        continue
    if not lower:
        print("One digit is required")
        continue
    if space:
        print("Space is not allowed")
        continue
    if not special:
        print("One special character is required")
        continue    
    valid = True
    if valid:
        print("Password Validation is successful!!")
        break
```

```python
# input : Apple green bananaa aabbcc helllo hello
# output : aple gren banana abc helo helo
s3 = "Apple green bananaa aabbcc helllo hello"
ans = ""
for i in s3.split():
    for j in range(1, len(i)):
        if j == 1:
            ans += i[j-1]
            continue
        if i[j] == i[j-1]:
            continue
        ans += i[j]
    ans += " "
print(ans)
    
```
