### Scores.txt Tasks (questions only)

**TASK 1: Season Summary**
Print the following, in this order:
1. Distinct teams (number of different team names, as Team A or Team B)
2. Average total points per game (both teams' scores added together)
3. Highest-scoring game (largest combined score; if tied, the first one in the file)
4. Closest game (smallest score difference; if tied, the first one in the file)

```python
data = []
with open("scores.txt") as f:
    for line in f:
        parts = line.split()
        data.append([parts[0], parts[1], int(parts[2]), parts[3], int(parts[4])])
teams = set()
points = []
highest = 0
margin = 1000
for date, team1, ts1, team2, ts2 in data:
    teams.add(team1)
    teams.add(team2)
    points.append(ts1+ts2)
    if ts1+ts2 > highest:
        highest = ts1+ts2
        h_date, h_team1, h_ts1, h_team2, h_ts2 = date, team1, ts1, team2, ts2
    if ts1 > ts2:
        margins = ts1-ts2
        if margins < margin:
            margin = margins
            m_date, m_team1, m_ts1, m_team2, m_ts2 = date, team1, ts1, team2, ts2
    else:
        margins = ts2-ts1
        if margins < margin:
            margin = margins
            m_date, m_team1, m_ts1, m_team2, m_ts2 = date, team1, ts1, team2, ts2

print("Distinct teams:", len(teams))
print("Average total points per game:", sum(points)/len(points))
print("Highest-scoring game:")
print(h_date, h_team1, h_ts1, h_team2, h_ts2)
print("Closest game:")
print(m_date, m_team1, m_ts1, m_team2, m_ts2)
```

**TASK 2: Team Performance Table (Top 10 teams by win percentage, minimum 10 games)**
For each team, find games played, wins, losses, and average margin of victory. Average margin is (total points scored minus total points allowed) divided by games played.

Include only teams with at least 10 games. Sort by win percentage from highest to lowest; if two teams are tied, the team with more wins comes first. If teams are still tied, the team that appears first in the file comes first. Print the top 10 teams. Win percentage is (wins / games played) x 100.

```python
played = {}
win = {}
loses = {}
records = {}
my_score = {}
opp_score = {}
winning_percent = []
for date, team1, ts1, team2, ts2 in data:
    played[team1] = played.get(team1, 0)+1
    played[team2] = played.get(team2, 0)+1
    my_score[team1] = my_score.get(team1, 0) + ts1
    opp_score[team1] = opp_score.get(team1, 0) + ts2
    my_score[team2] = my_score.get(team1, 0) + ts1
    opp_score[team2] = opp_score.get(team2, 0) + ts2
    if ts1 > ts2:
        win[team1] = win.get(team1, 0)+1
        loses[team1] = loses.get(team2, 0)+1
    else:
        win[team2] = win.get(team2, 0)+1
        loses[team1] = loses.get(team1, 0)+1
records = {i: {"Played": played.get(i, 0), "Win": win.get(i, 0), "Loss": loses.get(i, 0), "My_Score": my_score.get(i, 0), "Opp_Score": opp_score.get(i, 0)} for i in teams}   
for i in records:
    winning_percent.append([i, records[i]["Played"], (records[i]["Win"]/records[i]["Played"])*100, (records[i]["My_Score"]-records[i]["Opp_Score"])/records[i]["Played"]])
top = sorted(winning_percent, key = lambda x:x[2], reverse = True)
winning_percent
```

**TASK 3: Biggest Wins (Top 10 biggest wins)**
For every game, find the winner and the winning margin. Sort all games by margin from largest to smallest and print the top 10. Each line shows the rank, winner and score, a dash, loser and score, and the margin. If margins are tied, keep the order in which the games appear in the file. Each game gets its own rank, even when margins are tied.

```python
game = []
for date, team1, ts1, team2, ts2 in data:
    if ts1 > ts2:
        game.append([team1, ts1, team2, ts2, ts1-ts2])
    else:
        game.append([team2, ts2, team1, ts1, ts2-ts1])
task3 = sorted(game, key = lambda x:x[4], reverse = True)
rank = 0
for winner, win_score, loser, loser_score, margin in task3[0:10]:
    rank += 1
    print(rank, winner, win_score, loser, loser_score, margin)
```

**TASK 4: Monthly Scoring Trend**
Group games by month for November, December, January, February, March, and April, in that order. For each month, print the number of games and the average combined score per game. At the end, print the month with the highest average combined score and that average. If tied, report the earlier month.

**OUTPUT REQUIREMENTS**
- Follow the reference output exactly: line order, section dividers (60 "=" characters), labels, capitalization, spacing, column alignment, and blank lines.
- Printed headings, labels, and column names must match the reference output exactly, including the Task 4 header row and the "Highest average:" line.
- Team names and dates must appear exactly as in the file.
- Any difference from the reference output is marked incorrect, even if the numbers are correct.

```python
month = {11:[], 12: [], 1: [], 2:[], 3:[], 4:[]}
for date, team1, ts1, team2, ts2 in data:
    m = int(date.split('/')[0])
    month[m].append(ts1+ts2)
month_records = []
for i in month:
    month_records.append([i, len(month[i]), sum(month[i]), sum(month[i])/len(month[i])])
print(month_records) 
maximum = max(month_records, key = lambda x:x[-1])
print("Maximum Average", maximum)
```
