```python
# to check whether the entered string is binary or not
s = "aaabbb"
print(len(set(s)) == 2)
```

```python
s = {{1, 2}, {1, 2}} # not possible as set is mutable but set contains only immutable elements
```

```python
s = {(1, 2), (3, 4)} # possible as tuple is immutable
print(s)
```

```python
s1 = {1, 2, 3}
s2 = {4, 5, 2, 3}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s1.symmetric_difference(s2)) # (s1 union s2) - (s1 intersection s2)
print()
print(s1 | s2)
print(s1 & s2)
print(s1 - s2)
print(s1 ^ s2)
```

```python
s1 = {1, 2, 3}
s2 = {1, 2}
s2.issubset(s1)
s1.issuperset(s2)
```
